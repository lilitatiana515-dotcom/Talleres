Inteligencia Artificial II

En el video se explicó el tema del aprendizaje automático o Machine Learning, donde aprendimos sobre diferentes herramientas y tecnologías que se utilizan para trabajar con datos y desarrollar modelos de inteligencia artificial. Entre ellas se mencionaron Python Data Stack, Scikit-learn Suite, entre otras.

Uno de los temas que me pareció importante fue la fuga de datos. Esta sucede cuando información que pertenece al conjunto de prueba termina siendo utilizada durante el entrenamiento del modelo. Esto puede hacer que el modelo parezca tener mejores resultados de los que realmente tiene y, por lo tanto, la evaluación deja de ser confiable.

También se explicaron diferentes procesos que se realizan antes o durante la creación de un modelo, como la imputación de datos, que sirve para completar valores faltantes; el escalamiento, que permite llevar los datos a una escala similar; la reducción de dimensionalidad, que ayuda a disminuir la cantidad de variables; y los clasificadores, que permiten asignar los datos a diferentes categorías.

Otro concepto explicado fue el truco del kernel , utilizado en algoritmos como SVM. Esta técnica permite transformar los datos a un espacio de mayor dimensión para poder solucionar problemas que no pueden separarse fácilmente mediante una línea o una frontera simple.

También vimos dos técnicas importantes de aprendizaje conjunto:

Bagging: combina varios modelos para obtener mejores resultados y reducir la variabilidad. Un ejemplo es Random Forest, que utiliza varios árboles de decisión.
Boosting: crea los modelos de forma consecutiva, haciendo que cada nuevo modelo intente corregir los errores cometidos por el anterior.

Además, se explicó el concepto de sesgo y varianza . Este consiste en encontrar un equilibrio entre un modelo demasiado simple, que puede no aprender correctamente los datos, y uno demasiado complejo, que puede llegar a memorizar el ruido de los datos y funcionar mal con información nueva.

Para encontrar una mejor configuración de los modelos se mencionaron GridSearchCV y RandomizedSearchCV.
- GridSearchCV prueba diferentes combinaciones de parámetros de manera sistemática, mientras que -RandomizedSearchCV prueba diferentes configuraciones de forma aleatoria, lo que puede ahorrar tiempo cuando existen muchas posibilidades.

Finalmente, se estudiaron diferentes métricas para evaluar los modelos de clasificación, como la matriz de confusión, la precisión , el F1-Score, las curvas ROC y el área bajo la curva . Estas métricas permiten conocer qué tan bien está funcionando un modelo y cuáles son sus errores al realizar las predicciones.
